## Abbreviation
    . We count manually replace every country but it will take time, So I looked up for library.
    . Found a library called "countrycode"
    . In our data like  RSA Republic of South Africa are there let's clearn that up, so that library does not throw warning
    . Here few data are of Islands, let's group that to different category, European Community, Channel Islands and Unspecified to other categories
    . On one side we have filtered based on known name of country "specified_country" and in other where there is islands and another stuff we added them into seprate data frame "specified_country"

```r
library(countrycode)
library(dplyr) # for that pipe opreator and %in% opreator: incase I forget later, %in% is used to check membership of one vector is found in another

df$Country[df$Country == "RSA"] <- "South Africa"
df$Country[df$Country == "EIRE"] <- "Ireland"



unspecified_country <- df %>%
  filter(Country ==  "European Community" | Country == "Channel Islands" | Country == "Unspecified")

specified_country <- df[df$Country %in% c("European Community", "Channel Islands", "Unspecified") == FALSE, ]
# df$Country[df$Country == "Unspecified"] <- NA
country_codes <- countrycode(specified_country$Country, origin = "country.name", destination = "iso3c") # "country.name" Full English country name, destination is the format of our output, ISO 3-letter code
# df$Country <- country_codes
# Let's check the number of NA in country now, we need to know if that's large number or small
unique(unspecified_country$Country)
```
