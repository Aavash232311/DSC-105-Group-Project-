## Data Cleaning Plan

- Handle missing Customer IDs:
  - Replace with `"Unknown"`

```r
any(is.na(df$CustomerID)) # I am checking this because I know there is null, I loaded this csv into local database and when selecting "CustomerID" as a PK it throwed error
df$CustomerID[is.na(df$CustomerID)] <- "Unknown"
any(is.na(df$CustomerID))
```

## Abbreviation
    . We count manually replace every country but it will take time, So I looked up for library.
    . Found a library called "countrycode"
    . In our data like  RSA Republic of South Africa are there let's clearn that up, so that library does not throw warning
    . Here few data are of Islands, European Community so we will set that up to N/A

```r
library(countrycode)

df$Country[df$Country == "RSA"] <- "South Africa"
df$Country[df$Country == "EIRE"] <- "Ireland"
df$Country[df$Country == "European Community"] <- NA
df$Country[df$Country == "Channel Islands"] <- NA
df$Country[df$Country == "Unspecified"] <- NA
country_codes <- countrycode(df$Country, origin = "country.name", destination = "iso3c") # "country.name" Full English country name, destination is the format of our output, ISO 3-letter code
df$Country <- country_codes
df
# Let's check the number of NA in country now, we need to know if that's large number or small
number_of_na_country <- sum(is.na(df$Country))
paste("Total country", length(df$Country))
paste("N/A country record", number_of_na_country)
```
